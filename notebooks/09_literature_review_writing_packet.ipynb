# Literature Review Writing Packet

This notebook demonstrates a writing packet for one synthetic theme. The packet brings together section outline, evidence matrix, claim bank, citation bank, paragraph plan, and readiness report so the user can decide what is safe to draft.

The packet does not write a finished literature-review section.

```python
from pathlib import Path

from paper_workbench.authoring import writing_packet_report
from paper_workbench.bibtex import parse_bibtex_file
from paper_workbench.claims import collect_notes
from paper_workbench.registry import load_registry
from paper_workbench.tags import load_themes

project = Path('projects/zis_photocatalysis')
papers = load_registry(project / 'registry.csv')
notes = collect_notes(project / 'notes')
claims = [claim for note in notes for claim in note.claims]
entries = parse_bibtex_file(project / 'bibtex/library.bib')
themes = load_themes(project / 'themes.json')

packet = writing_packet_report('photocorrosion', papers, notes, claims, entries, themes, project='zis_photocatalysis')
print(packet[:2500])
```

The photocorrosion fixture intentionally has weak evidence and a missing evidence location. The packet should make that visible before drafting.

```python
assert 'does not fabricate claims' in packet
assert 'Claims Not Ready' in packet
assert 'missing evidence' in packet.lower()
print('Packet boundary and weak-evidence diagnostics are present.')
```

## Key Takeaways

- A writing packet is a pre-drafting audit bundle.
- It should surface weak claims and missing evidence before writing starts.
- It is not a substitute for reading papers or writing the final section yourself.
