# Evidence Matrix Workflow

This notebook shows how a researcher can turn structured notes into a theme-specific evidence matrix. The workflow matters because literature-review claims should be traceable to papers, evidence types, and evidence locations before drafting begins.

All data used here is synthetic.

```python
from pathlib import Path

from paper_workbench.authoring import build_evidence_matrix, evidence_matrix_report
from paper_workbench.claims import collect_notes
from paper_workbench.registry import load_registry
from paper_workbench.tags import load_themes

project = Path('projects/zis_photocatalysis')
papers = load_registry(project / 'registry.csv')
notes = collect_notes(project / 'notes')
claims = [claim for note in notes for claim in note.claims]
themes = load_themes(project / 'themes.json')

len(papers), len(notes), len(claims), [theme.theme_id for theme in themes]
```

The matrix is built from claims that were already entered in structured notes. It does not infer missing claims from paper titles or abstracts.

```python
matrix = build_evidence_matrix('charge separation', papers, claims, themes, notes, project='zis_photocatalysis')
print(evidence_matrix_report(matrix))
```

## Key Takeaways

- Evidence matrices help verify which claims have paper IDs, BibTeX keys, confidence values, and evidence locations.
- Missing fields should be fixed in notes before the claim is cited.
- The report is a planning artifact, not final prose.
