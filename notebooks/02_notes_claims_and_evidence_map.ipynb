# Notes Claims and Evidence Map

Synthetic example showing note-template rendering, note parsing, claim extraction, theme grouping, and evidence-map generation.

```python
from pathlib import Path

root = Path.cwd()
if not (root / "data").exists():
    root = root.parent

registry_path = root / "data" / "registries" / "example_papers.csv"
notes_dir = root / "data" / "notes"
themes_path = root / "data" / "examples" / "themes.json"
```

```python
from paper_workbench.registry import load_registry
from paper_workbench.notes import render_note_template, parse_note_file

papers = load_registry(registry_path)
template_preview = render_note_template(papers[0]).splitlines()[:12]
parsed_note = parse_note_file(notes_dir / "example_note_1.md")

template_preview, parsed_note.paper_id, parsed_note.claims[0].claim_text
```

```python
from paper_workbench.claims import collect_claims
from paper_workbench.tags import load_themes, group_claims_by_theme
from paper_workbench.reporting import evidence_map_report

claims = collect_claims(notes_dir)
themes = load_themes(themes_path)
grouped = group_claims_by_theme(claims, themes)
evidence_map = evidence_map_report(papers, claims, themes)

len(claims), sorted(grouped), evidence_map.splitlines()[:10]
```
