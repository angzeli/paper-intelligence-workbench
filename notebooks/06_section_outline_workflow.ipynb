# Section Outline Workflow

A section outline turns tracked evidence into a writing plan for one literature-review subsection. It is deliberately not polished prose: it lists papers, claims, weak evidence, unresolved questions, and missing evidence to address before drafting.

This notebook uses only the synthetic `zis_photocatalysis` project.

```python
from pathlib import Path
import sys

root = Path.cwd()
if not (root / "paper_workbench").exists():
    root = root.parent
sys.path.insert(0, str(root))
```

## Load the Project Evidence

The outline should be built only from user-entered notes and claims. Nothing is inferred from paper titles alone.

```python
from paper_workbench.claims import collect_claims, collect_notes
from paper_workbench.projects import load_project_profile
from paper_workbench.registry import load_registry
from paper_workbench.tags import load_themes

profile = load_project_profile("zis_photocatalysis", root=root)
papers = load_registry(profile.registry_path)
notes = collect_notes(profile.notes_dir)
claims = collect_claims(profile.notes_dir)
themes = load_themes(profile.themes_path)

[(paper.paper_id, paper.title, paper.bibtex_key) for paper in papers]
```

## Choose a Theme

For a real review, choose a theme only after confirming that notes and claims are populated. Here the synthetic project includes a `photocorrosion` theme.

```python
from paper_workbench.tags import group_claims_by_theme

theme_query = "photocorrosion"
grouped = group_claims_by_theme(claims, themes)
[(claim.claim_id, claim.strength, claim.section or claim.page or "missing evidence") for claim in grouped.get(theme_query, [])]
```

## Build the Outline

The outline pulls from tracked claims and note fields. It should make gaps visible before you start writing narrative prose.

```python
from paper_workbench.reporting import section_outline_report

outline = section_outline_report(theme_query, papers, claims, themes, notes)
outline.splitlines()[:20]
```

## Inspect Weak Claims and Missing Evidence

A useful outline should tell you what not to rely on yet.

```python
lines = outline.splitlines()
[line for line in lines if "weak" in line.lower() or "missing evidence" in line.lower() or "add evidence" in line.lower()]
```

## Check Citation Keys

The outline lists BibTeX keys to include, but only if registry rows have verified keys.

```python
[line for line in lines if "zisStability2024" in line]
```

## Key Takeaways

- Section outlines are evidence checklists, not generated literature-review prose.
- Weak claims and missing evidence locations should be resolved before drafting.
- Citation keys come from the registry and BibTeX workflow; they are not invented.
- The outline is most useful after running citation audit and evidence-map reports.
