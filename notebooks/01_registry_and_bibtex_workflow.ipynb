# Registry and BibTeX Workflow

Synthetic example showing registry loading, validation, duplicate detection, BibTeX validation, and audit-report generation.

```python
from pathlib import Path

root = Path.cwd()
if not (root / "data").exists():
    root = root.parent

registry_path = root / "data" / "registries" / "example_papers.csv"
bibtex_path = root / "data" / "bibtex" / "example_library.bib"
```

```python
from paper_workbench.registry import load_registry, validate_registry, detect_duplicate_doi, detect_duplicate_title

papers = load_registry(registry_path)
registry_findings = validate_registry(papers)

len(papers), [finding.code for finding in registry_findings], detect_duplicate_doi(papers), detect_duplicate_title(papers)
```

```python
from paper_workbench.bibtex import parse_bibtex_file, validate_bibtex
from paper_workbench.reporting import bibtex_audit_report

entries = parse_bibtex_file(bibtex_path)
bib_findings = validate_bibtex(entries, papers)
report_markdown = bibtex_audit_report(entries, bib_findings)

len(entries), [finding.code for finding in bib_findings[:5]], report_markdown.splitlines()[:6]
```
