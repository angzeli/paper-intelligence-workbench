# Subsection Readiness Workflow

This notebook shows how to calculate a local completeness score for a theme. The score helps decide whether a subsection is ready to outline or still needs targeted follow-up.

The score is not a truth score.

```python
from pathlib import Path

from paper_workbench.authoring import build_subsection_readiness, subsection_readiness_report
from paper_workbench.bibtex import parse_bibtex_file
from paper_workbench.claims import collect_notes
from paper_workbench.registry import load_registry
from paper_workbench.tags import load_themes

project = Path('projects/zis_photocatalysis')
papers = load_registry(project / 'registry.csv')
notes = collect_notes(project / 'notes')
claims = [claim for note in notes for claim in note.claims]
entries = parse_bibtex_file(project / 'bibtex/library.bib')
themes = load_themes(project / 'themes.json')
```

```python
for theme in ['charge separation', 'photocorrosion']:
    readiness = build_subsection_readiness(theme, papers, notes, claims, entries, themes, project='zis_photocatalysis')
    print(subsection_readiness_report(readiness))
```

## Key Takeaways

- Readiness scoring rewards complete local evidence tracking.
- Missing notes, missing evidence locations, and missing BibTeX links reduce readiness.
- The score should guide follow-up reading and note cleanup, not replace judgment.
